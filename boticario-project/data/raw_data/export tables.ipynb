```python
df_sku = pd.read_excel("base-sortimento-campanha.xlsx", sheet_name="tb_sku")
df_ciclo = pd.read_excel("base-sortimento-campanha.xlsx", sheet_name="tb_ciclo")
df_venda = pd.read_excel("base-sortimento-campanha.xlsx", sheet_name="tb_venda")
```

## Join tables

```python
df_venda_j1 = df_venda.join(df_loja.set_index('cod_loja'), on = 'cod_loja', how='left')
df_venda_j1.head(5)
```

```python
df_venda_j2 = df_venda_j1.join(df_sku.set_index('cod_sku'), on = 'cod_sku', how='left')
df_venda_j2.head(5)
```

```python
del df_venda_j1
```

```python
df_venda_final = df_venda_j2.join(df_ciclo.set_index('cod_ciclo'), on = 'cod_ciclo', how='left')
df_venda_final.head(5)
```

```python
del df_venda_j2
```

## Export table in csv

```python
df_venda_final.to_csv("../processed_data/df_venda_final.csv", index=False)
```
