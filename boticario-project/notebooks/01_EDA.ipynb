```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Data load

### Base agregada de vendas (qtd vendida) por loja, sku, tipo de loja e que traz o espaco disponivel na gondola

```python
df = pd.read_csv('../data/processed_data/df_venda_final.csv')

df.head(3)
```

```python
code_variables = ['cod_loja', 'cod_sku', 'cod_ciclo', 'cod_ciclo_inicio']
numeric_cols = df.drop(code_variables, axis=1).select_dtypes(include=['float64', 'int64']).columns
categorical_cols = df.drop(code_variables, axis=1).select_dtypes(include=['object', 'category']).columns
all_wtithout_code_variables = df.drop(code_variables, axis=1).columns
```

## Iterating box plots of numerical

```python
cols_to_ignore = list(code_variables) + ['descricao_sku', 'sku_description']  # ajuste para o nome exato da sua coluna

categorical_cols_graph = (
    df.drop(columns=cols_to_ignore, errors='ignore')
    .select_dtypes(include=['object', 'category'])
    .columns
    .tolist()
)

for cat in categorical_cols:
    for num in numeric_cols:
        plt.figure(figsize=(10, 5))
        
        sns.boxplot(
            data=df,
            x=cat,
            y=num,
            palette='coolwarm',
            showmeans=True,  # adiciona o ponto da média para complementar a mediana
            meanprops={"marker": "o", "markerfacecolor": "white", "markeredgecolor": "black"}
        )
        
        plt.title(f'Distribuição de {num} por {cat}', fontsize=12, pad=10)
        plt.xlabel(cat, fontsize=10)
        plt.ylabel(num, fontsize=10)
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
```

## Ploting sample stores and most selled item

```python
for i in range(len(stores_rnd)):
    df_temp = df.query(f'cod_loja == "{stores_rnd[i]}" and cod_sku == "S0040"').reset_index()
    df_temp['cod_ciclo'] = df_temp['cod_ciclo'].astype(str)  # Convertendo para string para evitar problemas de plotagem
    df_temp = df_temp.set_index('cod_ciclo')
    plt.figure(figsize=(12, 6))
    sns.lineplot(df_temp['qt_venda'])
    plt.title(f'Evolução de Vendas para Loja {stores_rnd[i]} e SKU S0040', fontsize=14)
    plt.show()
```
